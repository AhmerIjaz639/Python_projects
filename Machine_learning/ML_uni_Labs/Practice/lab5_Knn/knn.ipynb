```python
# ─────────────────────────────────────────
# STEP 1: Libraries Import Karo
# ─────────────────────────────────────────
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn import datasets



# ─────────────────────────────────────────
# STEP 2: Data Load Karo
# ─────────────────────────────────────────
iris = datasets.load_iris()


x = iris.data      # Features (sepal length, sepal width, petal length, petal width)
y = iris.target    # Labels (0, 1, 2 → 3 types of flowers)

print("x shape:", x.shape)   # (150, 4) → 150 flowers, 4 features each
print("y shape:", y.shape)   # (150,)   → 150 labels

# ─────────────────────────────────────────
# STEP 3: Data Split Karo (Training + Testing)
# ─────────────────────────────────────────
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3
)


# ─────────────────────────────────────────
# STEP 4: Model Banao aur Train Karo
# ─────────────────────────────────────────
classifier = KNeighborsClassifier(n_neighbors=5)


classifier.fit(x_train, y_train)


# ─────────────────────────────────────────
# STEP 5: Prediction Karo
# ─────────────────────────────────────────
y_pred = classifier.predict(x_test)


# ─────────────────────────────────────────
# STEP 6: Results Check Karo
# ─────────────────────────────────────────
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```
