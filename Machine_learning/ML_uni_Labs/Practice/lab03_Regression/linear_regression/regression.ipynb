```python
import pandas as pd 
import matplotlib.pyplot as plt
data={
    "YOE":[1,2,3,4,5,6,7,8],
    "salary":[1000,20000,50000,80000,90000,100000,140000,200000]
}

#plt.plot(data["YOE"], data["salary"], marker='o', color='b', linestyle='-')


df=pd.DataFrame(data)
print(df)

X=df[["YOE"]] #use [[]]  becasuse hum is me predict hone wala data ko 2d me dete hein 
Y=df["salary"]

from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X,Y)

prediction = model.predict([[7]])


print("prediction:",prediction)
print("Slope:", model.coef_)
print("Intercept:", model.intercept_)
```

```python
import pandas as pd 
import matplotlib.pyplot as plt
data = {
    "YOE": [
        1.1, 1.3, 1.5, 2.0, 2.2, 2.9, 3.0, 3.2, 3.2, 3.7, 
        4.0, 4.5, 5.1, 5.9, 6.0, 6.8, 7.1, 8.2, 9.0, 10.5
    ],
    "salary": [
        39343, 46205, 37731, 43525, 39891, 56642, 60150, 54445, 64445, 57189,
        63218, 61111, 66029, 81363, 93940, 91738, 98273, 113812, 105582, 122391
    ]
}


#plt.plot(data["YOE"], data["salary"], marker='o', color='b', linestyle='-')


df=pd.DataFrame(data)
print(df)

X=df[["YOE"]]
Y=df["salary"]

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,Y,test_size=0.25,random_state=10)
model=LinearRegression()
model.fit(X_train,y_train)

y_pred= model.predict(X_test)


print("prediction:",y_pred)
print("actuall:",y_test.values)
print("r score:" , model.score(X_test,y_test))
```

```python
import pandas as pd 
import matplotlib.pyplot as plt
data = {
    "YOE": [
        1.1, 1.3, 1.5, 2.0, 2.2, 2.9, 3.0, 3.2, 3.2, 3.7, 
        4.0, 4.5, 5.1, 5.9, 6.0, 6.8, 7.1, 8.2, 9.0, 10.5
    ],
    "salary": [
        39343, 46205, 37731, 43525, 39891, 56642, 60150, 54445, 64445, 57189,
        63218, 61111, 66029, 81363, 93940, 91738, 98273, 113812, 105582, 122391
    ]
}

#plt.plot(data["YOE"], data["salary"], marker='o', color='b', linestyle='-')

df=pd.DataFrame(data)
#print(df)

X=df[["YOE"]]                                                  #use [[]]  becasuse hum is me predict hone wala data ko 2d me dete hein 
Y=df["salary"]

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_absolute_error

X_train,X_test,y_train,y_test=train_test_split(
    X,Y,test_size=0.25,random_state=10
    )

model=LinearRegression()
model.fit(X_train,y_train)

y_pred= model.predict(X_test)

print("compare actuall train data and test pred  data r^2  :",r2_score(y_test,y_pred))
print("tells how well your model predictions match real data  r^2 :",model.score(X_train,y_train))

print("prediction:",y_pred)
print("actuall:",y_test.values)


#mean absolute error 
mae=mean_absolute_error(y_test,y_pred)
print("MAE tells how far is your model predcitions average from actual values: ", mae)
```
