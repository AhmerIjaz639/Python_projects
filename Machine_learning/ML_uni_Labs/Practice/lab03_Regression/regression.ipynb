```python
import pandas as pd
data = {
    "name": ["spongebob", "patric", "squidward"],
    "age": [30, 20, 40]
}

df = pd.DataFrame(data)

new_row = pd.DataFrame({"name": ["Charlie"], "age": [28]})
df=pd.concat([df, new_row], ignore_index=True)

df["JOB"]=["ai","n/a","cartoon","captain"]
print(df)
```

```python
import pandas as pd
df=pd.read_csv("data.csv")
order_no = input("Enter order_id to get details: ")

try:
    order_no = int(order_no)
    order = df[df["order_id"] == order_no]

    if order.empty:
        print("Order ID not found!")
    else:
        print(order.to_string(index=False))

except ValueError:
    print("Order ID not found!")
```
