```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Load Dataset
iris = load_iris()

X = iris.data
y = iris.target

# Train Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.4, random_state=0
)

# Model
model = GaussianNB()

# Train
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Evaluation
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Categories
categories = [
    'talk.religion.misc',
    'soc.religion.christian',
    'sci.space',
    'comp.graphics'
]

# Load Dataset
train = fetch_20newsgroups(subset='train', categories=categories)
test = fetch_20newsgroups(subset='test', categories=categories)

# Model
model = make_pipeline(
    TfidfVectorizer(),
    MultinomialNB()
)

# Train
model.fit(train.data, train.target)

# Predict
y_pred = model.predict(test.data)

# Evaluation
print("Accuracy:", accuracy_score(test.target, y_pred))
print("Confusion Matrix:")
print(confusion_matrix(test.target, y_pred))
print("Classification Report:")
print(classification_report(test.target, y_pred))


texts = [
    "NASA launched a rocket",
    "Jesus Christ and Christianity",
    "Computer graphics and screen resolution",
    "Discussion about religion and atheism"
]

predictions = model.predict(texts)

for i in range(len(texts)):
    print(texts[i], "→", train.target_names[predictions[i]])
```
