```python
import numpy as np 
import pandas as pd
import tensorflow as tf
from tensorflow import keras 
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import os 

df=pd.read_csv("insurance.csv")

X_train,X_test,y_train,y_test=train_test_split(
    df[['age','affordibility']],df['bought_inusrance'],test_size=0.2,random_state=25)
X_train['age']=X_train['age']/100
X_test['age']=X_test['age']/100

model=keras.Sequential(
    [keras.layers.Dense(1,input(2,),activation='sigmoid',
                        kernel_initializer='ones',bias_initializer='zeros')]
)

model.compile(optimizer='adam',loss='binary_crossentropy',metrics=['accuracy'])

model.fit(X_train,y_train,epochs=5000,verbose=0)

print(model.evaluate(X_test,y_test))
print(model.predict(X_test))

coef,intercept=model.get_weights()
print("w1=",coef[0][0],"w2=",coef[1][0],"bias=",intercept[0])
```
