```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([2023,2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3 = np.array([13, 15, 20, 30])

plt.title("Class size",fontsize=20,family="Arial")
plt.xlabel("Year", fontsize=20,
family="Arial",
fontweight="bold",
color="#2dbefc")

plt.ylabel("Students", fontsize=20,
family="Arial",
fontweight="bold",
color="#2dbefc")
plt.tick_params(axis="both",
colors="#5c3304")

plt.plot(x,y1)
plt.plot(x, y2)
plt.plot(x, y3)
plt.xticks(x)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# grid() = Helps make plots easier to read

x = [1, 2, 3, 4, 5]
y= [5, 10, 15, 20, 25]

plt.grid( #axis="y" or "x"
linewidth=2,
color="lightgray",
linestyle="dashdot")

plt.plot(x, y)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Bar chart = compare categories of data by representing each category with a

categories = np.array(["Grains", "Fruit", "Vegetables", "Protein", "Dairy", "Sweets"])
values = np.array([4, 3, 2, 5, 3, 1])

plt.bar(categories, values)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Bar chart = Circular chart divided into slices to show percentages of the total.
#Good for visualizing distribution among categories.

categories = ["Freshmen", "Sophomores", "Juniors", "Seniors"]
values = np.array([300, 250, 275, 225])

plt.pie(values, labels=categories,
autopct="%1.1f%%")

plt.title("Bro Code College")

plt.show()

```
